```python
# PART-1 
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score
from PIL import Image
from transformers import AutoModel, AutoFeatureExtractor
import matplotlib.pyplot as plt
from tqdm import tqdm
import warnings
import os
import re
warnings.filterwarnings('ignore')

# ================== DATA PREPARATION ==================

class CheXpertLabeler:
    """
    Simplified CheXpert labeler to extract findings from reports
    """
    def __init__(self):
        # Define patterns for common findings
        self.finding_patterns = {
            'Cardiomegaly': r'(cardiomegaly|enlarged heart|cardiac enlargement)',
            'Edema': r'(edema|pulmonary edema|interstitial edema)',
            'Consolidation': r'(consolidation|consolidative)',
            'Atelectasis': r'(atelectasis|atelectatic)',
            'Pleural Effusion': r'(pleural effusion|effusion)',
            'Pneumonia': r'(pneumonia|pneumonic)',
            'Pneumothorax': r'(pneumothorax)',
            'No Finding': r'(normal|no acute|unremarkable|clear)'
        }
        
        # Negation patterns
        self.negation_patterns = r'(no|without|absent|negative|denied|resolved|not)'
    
    def extract_labels(self, report_text):
        """Extract binary labels from report text"""
        if pd.isna(report_text):
            return {finding: 0 for finding in self.finding_patterns.keys()}
        
        report_text = report_text.lower()
        labels = {}
        
        for finding, pattern in self.finding_patterns.items():
            # Check if finding is mentioned
            match = re.search(pattern, report_text)
            if match:
                # Check for negation before the finding
                start = max(0, match.start() - 50)
                preceding_text = report_text[start:match.start()]
                if re.search(self.negation_patterns, preceding_text):
                    labels[finding] = 0
                else:
                    labels[finding] = 1
            else:
                labels[finding] = 0
        
        # If no specific findings, mark as "No Finding"
        if all(v == 0 for k, v in labels.items() if k != 'No Finding'):
            labels['No Finding'] = 1
        else:
            labels['No Finding'] = 0
            
        return labels

def prepare_multiview_data(df1, df2, image_dir):
    """
    Prepare multi-view dataset with frontal and lateral pairs
    """
    multiview_data = []
    labeler = CheXpertLabeler()
    
    # Group images by patient
    patient_groups = df2.groupby('uid')
    
    for uid, group in patient_groups:
        # Get report for this patient
        report_row = df1[df1['uid'] == uid]
        if report_row.empty:
            continue
            
        findings = report_row['findings'].values[0]
        if pd.isna(findings):
            continue
        
        # Extract labels from report
        labels = labeler.extract_labels(findings)
        
        # Get frontal and lateral views
        frontal_imgs = group[group['projection'].str.contains('Frontal', case=False, na=False)]
        lateral_imgs = group[group['projection'].str.contains('Lateral', case=False, na=False)]
        
        if not frontal_imgs.empty and not lateral_imgs.empty:
            # Take first frontal and lateral pair
            frontal_path = os.path.join(image_dir, frontal_imgs.iloc[0]['filename'])
            lateral_path = os.path.join(image_dir, lateral_imgs.iloc[0]['filename'])
            
            multiview_data.append({
                'uid': uid,
                'frontal_path': frontal_path,
                'lateral_path': lateral_path,
                'findings': findings,
                'labels': labels
            })
    
    return pd.DataFrame(multiview_data)



# ================== MODEL ARCHITECTURE ==================

class BidirectionalCrossAttention(nn.Module):
    """
    Bidirectional Cross-Attention module for fusing frontal and lateral views
    """
    def __init__(self, hidden_dim, num_heads=8, dropout=0.1):
        super().__init__()
        self.num_heads = num_heads
        self.hidden_dim = hidden_dim
        self.head_dim = hidden_dim // num_heads
        
        assert hidden_dim % num_heads == 0, "hidden_dim must be divisible by num_heads"
        
        # Cross-attention layers (frontal attending to lateral)
        self.f2l_attention = nn.MultiheadAttention(
            embed_dim=hidden_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )
        
        # Cross-attention layers (lateral attending to frontal)
        self.l2f_attention = nn.MultiheadAttention(
            embed_dim=hidden_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )
        
        # Feed-forward networks
        self.ffn_frontal = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim * 4),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim * 4, hidden_dim),
            nn.Dropout(dropout)
        )
        
        self.ffn_lateral = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim * 4),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim * 4, hidden_dim),
            nn.Dropout(dropout)
        )
        
        # Layer normalization
        self.norm1_f = nn.LayerNorm(hidden_dim)
        self.norm2_f = nn.LayerNorm(hidden_dim)
        self.norm1_l = nn.LayerNorm(hidden_dim)
        self.norm2_l = nn.LayerNorm(hidden_dim)
        
    def forward(self, frontal_features, lateral_features):
        """
        Args:
            frontal_features: [batch, seq_len, hidden_dim]
            lateral_features: [batch, seq_len, hidden_dim]
        Returns:
            enhanced_frontal: [batch, seq_len, hidden_dim]
            enhanced_lateral: [batch, seq_len, hidden_dim]
        """
        # Frontal attending to lateral
        f2l_out, _ = self.f2l_attention(
            query=frontal_features,
            key=lateral_features,
            value=lateral_features
        )
        frontal_enhanced = self.norm1_f(frontal_features + f2l_out)
        frontal_out = self.norm2_f(frontal_enhanced + self.ffn_frontal(frontal_enhanced))
        
        # Lateral attending to frontal
        l2f_out, _ = self.l2f_attention(
            query=lateral_features,
            key=frontal_features,
            value=frontal_features
        )
        lateral_enhanced = self.norm1_l(lateral_features + l2f_out)
        lateral_out = self.norm2_l(lateral_enhanced + self.ffn_lateral(lateral_enhanced))
        
        return frontal_out, lateral_out

class MultiViewCXRVisionModel(nn.Module):
    """
    Multi-view CXR Vision Model with Bidirectional Cross-Attention Fusion
    """
    def __init__(self, encoder_checkpoint="microsoft/swin-base-patch4-window7-224", 
                 num_labels=8, hidden_dim=768, num_fusion_layers=3):
        super().__init__()
        
        # Load pre-trained vision encoder (shared for both views)
        self.vision_encoder = AutoModel.from_pretrained(encoder_checkpoint)
        self.hidden_dim = hidden_dim
        
        # Projection layers to ensure consistent dimensions
        encoder_dim = self.vision_encoder.config.hidden_size
        self.frontal_projection = nn.Linear(encoder_dim, hidden_dim)
        self.lateral_projection = nn.Linear(encoder_dim, hidden_dim)
        
        # Stack of bidirectional cross-attention layers
        self.fusion_layers = nn.ModuleList([
            BidirectionalCrossAttention(hidden_dim) 
            for _ in range(num_fusion_layers)
        ])
        
        # Global pooling strategies
        self.pooling = nn.AdaptiveAvgPool1d(1)
        
        # Classification head for pre-training
        self.classifier = nn.Sequential(
            nn.Linear(hidden_dim * 2, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(0.2),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.GELU(),
            nn.Dropout(0.2),
            nn.Linear(hidden_dim // 2, num_labels)
        )
        
    def extract_patch_embeddings(self, pixel_values, view_type='frontal'):
        """Extract patch embeddings while preserving spatial information"""
        outputs = self.vision_encoder(pixel_values, output_hidden_states=True)
        
        # Get the last hidden state (patch embeddings)
        patch_embeddings = outputs.last_hidden_state  # [batch, seq_len, hidden_dim]
        
        # Project to consistent dimension
        if view_type == 'frontal':
            patch_embeddings = self.frontal_projection(patch_embeddings)
        else:
            patch_embeddings = self.lateral_projection(patch_embeddings)
            
        return patch_embeddings
    
    def fuse_multiview_features(self, frontal_features, lateral_features):
        """Apply bidirectional cross-attention fusion"""
        # Apply multiple layers of cross-attention
        for fusion_layer in self.fusion_layers:
            frontal_features, lateral_features = fusion_layer(frontal_features, lateral_features)
        
        return frontal_features, lateral_features
    
    def create_unified_embedding(self, frontal_features, lateral_features):
        """Create unified patient embedding from fused features"""
        # Global pooling over sequence dimension
        frontal_pooled = self.pooling(frontal_features.transpose(1, 2)).squeeze(-1)  # [batch, hidden_dim]
        lateral_pooled = self.pooling(lateral_features.transpose(1, 2)).squeeze(-1)  # [batch, hidden_dim]
        
        # Concatenate both view representations
        unified_embedding = torch.cat([frontal_pooled, lateral_pooled], dim=1)  # [batch, hidden_dim*2]
        
        return unified_embedding
    
    def forward(self, frontal_images, lateral_images):
        """
        Forward pass through the entire model
        """
        # Step 1: Extract patch embeddings for both views
        frontal_patches = self.extract_patch_embeddings(frontal_images, 'frontal')
        lateral_patches = self.extract_patch_embeddings(lateral_images, 'lateral')
        
        # Step 2: Apply bidirectional cross-attention fusion
        frontal_fused, lateral_fused = self.fuse_multiview_features(frontal_patches, lateral_patches)
        
        # Step 3: Create unified patient embedding
        unified_embedding = self.create_unified_embedding(frontal_fused, lateral_fused)
        
        # Step 4: Classification (for pre-training)
        logits = self.classifier(unified_embedding)
        
        return {
            'logits': logits,
            'unified_embedding': unified_embedding,
            'frontal_features': frontal_fused,
            'lateral_features': lateral_fused
        }

# ================== DATASET CLASS ==================

class MultiViewCXRDataset(Dataset):
    """Dataset for multi-view CXR images with labels"""
    def __init__(self, df, feature_extractor, label_columns):
        self.df = df
        self.feature_extractor = feature_extractor
        self.label_columns = label_columns
        
    def __len__(self):
        return len(self.df)
    
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        
        # Load images
        frontal_img = Image.open(row['frontal_path']).convert('RGB')
        lateral_img = Image.open(row['lateral_path']).convert('RGB')
        
        # Process images
        frontal_inputs = self.feature_extractor(images=frontal_img, return_tensors='pt')
        lateral_inputs = self.feature_extractor(images=lateral_img, return_tensors='pt')
        
        # Extract labels
        labels = torch.tensor([row['labels'][col] for col in self.label_columns], dtype=torch.float)
        
        return {
            'frontal_pixel_values': frontal_inputs['pixel_values'].squeeze(0),
            'lateral_pixel_values': lateral_inputs['pixel_values'].squeeze(0),
            'labels': labels
        }

# ================== TRAINING FUNCTIONS ==================

def train_epoch(model, dataloader, optimizer, criterion, device):
    """Train for one epoch"""
    model.train()
    total_loss = 0
    all_preds = []
    all_labels = []
    
    progress_bar = tqdm(dataloader, desc='Training')
    for batch in progress_bar:
        # Move to device
        frontal = batch['frontal_pixel_values'].to(device)
        lateral = batch['lateral_pixel_values'].to(device)
        labels = batch['labels'].to(device)
        
        # Forward pass
        outputs = model(frontal, lateral)
        loss = criterion(outputs['logits'], labels)
        
        # Backward pass
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        # Track metrics
        total_loss += loss.item()
        all_preds.append(torch.sigmoid(outputs['logits']).cpu().detach())
        all_labels.append(labels.cpu().detach())
        
        progress_bar.set_postfix({'loss': loss.item()})
    
    # Calculate AUC
    all_preds = torch.cat(all_preds)
    all_labels = torch.cat(all_labels)
    aucs = []
    for i in range(all_labels.shape[1]):
        try:
            auc = roc_auc_score(all_labels[:, i], all_preds[:, i])
            aucs.append(auc)
        except:
            aucs.append(0.5)
    
    return total_loss / len(dataloader), np.mean(aucs)

def validate(model, dataloader, criterion, device):
    """Validate the model"""
    model.eval()
    total_loss = 0
    all_preds = []
    all_labels = []
    
    with torch.no_grad():
        for batch in tqdm(dataloader, desc='Validating'):
            frontal = batch['frontal_pixel_values'].to(device)
            lateral = batch['lateral_pixel_values'].to(device)
            labels = batch['labels'].to(device)
            
            outputs = model(frontal, lateral)
            loss = criterion(outputs['logits'], labels)
            
            total_loss += loss.item()
            all_preds.append(torch.sigmoid(outputs['logits']).cpu())
            all_labels.append(labels.cpu())
    
    # Calculate AUC
    all_preds = torch.cat(all_preds)
    all_labels = torch.cat(all_labels)
    aucs = []
    for i in range(all_labels.shape[1]):
        try:
            auc = roc_auc_score(all_labels[:, i], all_preds[:, i])
            aucs.append(auc)
        except:
            aucs.append(0.5)
    
    return total_loss / len(dataloader), np.mean(aucs), aucs

# ================== MAIN TRAINING SCRIPT ==================

def main():
    # Configuration
    config = {
        'encoder_checkpoint': 'google/vit-base-patch16-224-in21k',
        'batch_size': 8,
        'learning_rate': 1e-4,
        'num_epochs': 10,
        'device': torch.device('cuda' if torch.cuda.is_available() else 'cpu'),
        'num_workers': 2,
        'hidden_dim': 768,
        'num_fusion_layers': 3
    }
    
    print(f"Using device: {config['device']}")
    
    # Load data
    print("Loading datasets...")
    df1 = pd.read_csv('/kaggle/input/chest-xrays-indiana-university/indiana_reports.csv')
    df2 = pd.read_csv('/kaggle/input/chest-xrays-indiana-university/indiana_projections.csv')
    image_dir = '/kaggle/input/chest-xrays-indiana-university/images/images_normalized/'
    
    # Prepare multi-view data with labels
    print("Preparing multi-view dataset...")
    multiview_df = prepare_multiview_data(df1, df2, image_dir)
    print(f"Found {len(multiview_df)} multi-view pairs")
    
    # Get label columns
    label_columns = list(multiview_df.iloc[0]['labels'].keys())
    num_labels = len(label_columns)
    print(f"Number of labels: {num_labels}")
    print(f"Labels: {label_columns}")
    
    # Split data
    train_df, val_df = train_test_split(multiview_df, test_size=0.2, random_state=42)
    print(f"Training samples: {len(train_df)}")
    print(f"Validation samples: {len(val_df)}")
    
    # Initialize feature extractor
    feature_extractor = AutoFeatureExtractor.from_pretrained(config['encoder_checkpoint'])
    
    # Create datasets
    train_dataset = MultiViewCXRDataset(train_df, feature_extractor, label_columns)
    val_dataset = MultiViewCXRDataset(val_df, feature_extractor, label_columns)
    
    # Create dataloaders
    train_loader = DataLoader(
        train_dataset, 
        batch_size=config['batch_size'],
        shuffle=True,
        num_workers=config['num_workers'],
        pin_memory=True
    )
    
    val_loader = DataLoader(
        val_dataset,
        batch_size=config['batch_size'],
        shuffle=False,
        num_workers=config['num_workers'],
        pin_memory=True
    )
    
    # Initialize model
    print("Initializing model...")
    model = MultiViewCXRVisionModel(
        encoder_checkpoint=config['encoder_checkpoint'],
        num_labels=num_labels,
        hidden_dim=config['hidden_dim'],
        num_fusion_layers=config['num_fusion_layers']
    ).to(config['device'])
    
    # Setup training
    optimizer = torch.optim.AdamW(model.parameters(), lr=config['learning_rate'])
    criterion = nn.BCEWithLogitsLoss()
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer, T_max=config['num_epochs']
    )
    
    # Training loop
    best_val_auc = 0
    train_losses = []
    val_losses = []
    train_aucs = []
    val_aucs = []
    
    for epoch in range(config['num_epochs']):
        print(f"\nEpoch {epoch+1}/{config['num_epochs']}")
        
        # Train
        train_loss, train_auc = train_epoch(
            model, train_loader, optimizer, criterion, config['device']
        )
        
        # Validate
        val_loss, val_auc, val_aucs_per_label = validate(
            model, val_loader, criterion, config['device']
        )
        
        # Update scheduler
        scheduler.step()
        
        # Track metrics
        train_losses.append(train_loss)
        val_losses.append(val_loss)
        train_aucs.append(train_auc)
        val_aucs.append(val_auc)
        
        print(f"Train Loss: {train_loss:.4f}, Train AUC: {train_auc:.4f}")
        print(f"Val Loss: {val_loss:.4f}, Val AUC: {val_auc:.4f}")
        
        # Print per-label AUCs
        for label, auc in zip(label_columns, val_aucs_per_label):
            print(f"  {label}: {auc:.3f}")
        
        # Save best model
        if val_auc > best_val_auc:
            best_val_auc = val_auc
            print(f"New best model! Saving...")
            
            # Save vision backbone and fusion weights only (not classifier)
            vision_state = {
                'vision_encoder': model.vision_encoder.state_dict(),
                'frontal_projection': model.frontal_projection.state_dict(),
                'lateral_projection': model.lateral_projection.state_dict(),
                'fusion_layers': model.fusion_layers.state_dict(),
                'config': {
                    'hidden_dim': config['hidden_dim'],
                    'num_fusion_layers': config['num_fusion_layers'],
                    'encoder_checkpoint': config['encoder_checkpoint']
                }
            }
            torch.save(vision_state, '/kaggle/working/multiview_vision_backbone.pth')
    
    # Plot training curves
    plt.figure(figsize=(12, 4))
    
    plt.subplot(1, 2, 1)
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Val Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    
    plt.subplot(1, 2, 2)
    plt.plot(train_aucs, label='Train AUC')
    plt.plot(val_aucs, label='Val AUC')
    plt.xlabel('Epoch')
    plt.ylabel('AUC')
    plt.title('Training and Validation AUC')
    plt.legend()
    
    plt.tight_layout()
    plt.savefig('/kaggle/working/training_curves.png')
    plt.show()
    
    print(f"\nTraining complete! Best validation AUC: {best_val_auc:.4f}")
    print("Vision backbone saved to: /kaggle/working/multiview_vision_backbone.pth")

if __name__ == "__main__":
    main()
```

```python
# PART-2 REPORT GENERATOR
#  multiview_report_generator_fixed.py
# Complete corrected PART-2 script — replace your existing file with this.

import os
import re
import warnings
from tqdm import tqdm

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt

# transformers imports (explicit BioGPT import to avoid AutoModel auto-factory issues)
from transformers import (
    AutoModel,
    AutoFeatureExtractor,
    AutoTokenizer,
    BioGptForCausalLM,
)
# BitsAndBytesConfig may or may not be available depending on transformers/bitsandbytes versions
try:
    from transformers import BitsAndBytesConfig  # available in newer transformers
    _HAS_BNB_CONFIG = True
except Exception:
    _HAS_BNB_CONFIG = False

# PEFT (LoRA)
try:
    from peft import LoraConfig, get_peft_model, TaskType, prepare_model_for_kbit_training
    _HAS_PEFT = True
except Exception:
    _HAS_PEFT = False

# Evaluation utilities
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
import evaluate

warnings.filterwarnings("ignore")

# -------------------------
# Phase-1 components (fixed)
# -------------------------

class BidirectionalCrossAttention(nn.Module):
    """Bidirectional cross-attention layer used in Phase 1 fusion"""
    def __init__(self, hidden_dim, num_heads=8, dropout=0.1):
        super().__init__()
        assert hidden_dim % num_heads == 0, "hidden_dim must be divisible by num_heads"
        self.f2l_attention = nn.MultiheadAttention(embed_dim=hidden_dim, num_heads=num_heads, dropout=dropout, batch_first=True)
        self.l2f_attention = nn.MultiheadAttention(embed_dim=hidden_dim, num_heads=num_heads, dropout=dropout, batch_first=True)

        self.ffn_frontal = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim * 4),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim * 4, hidden_dim),
            nn.Dropout(dropout),
        )

        self.ffn_lateral = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim * 4),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim * 4, hidden_dim),
            nn.Dropout(dropout),
        )

        self.norm1_f = nn.LayerNorm(hidden_dim)
        self.norm2_f = nn.LayerNorm(hidden_dim)
        self.norm1_l = nn.LayerNorm(hidden_dim)
        self.norm2_l = nn.LayerNorm(hidden_dim)

    def forward(self, frontal_features, lateral_features):
        f2l_out, _ = self.f2l_attention(query=frontal_features, key=lateral_features, value=lateral_features)
        frontal_enhanced = self.norm1_f(frontal_features + f2l_out)
        frontal_out = self.norm2_f(frontal_enhanced + self.ffn_frontal(frontal_enhanced))

        l2f_out, _ = self.l2f_attention(query=lateral_features, key=frontal_features, value=frontal_features)
        lateral_enhanced = self.norm1_l(lateral_features + l2f_out)
        lateral_out = self.norm2_l(lateral_enhanced + self.ffn_lateral(lateral_enhanced))

        return frontal_out, lateral_out


class PretrainedVisionEncoder(nn.Module):
    """
    Loads the Phase-1 visual backbone saved as a checkpoint dict containing:
      - config (hidden_dim, num_fusion_layers, encoder_checkpoint)
      - vision_encoder (state_dict)
      - frontal_projection (state_dict)
      - lateral_projection (state_dict)
      - fusion_layers (state_dict)
    The module exposes extract_features(...) which returns a unified embedding.
    """
    def __init__(self, checkpoint_path, device='cuda'):
        super().__init__()
        device = torch.device(device)
        checkpoint = torch.load(checkpoint_path, map_location=device)
        config = checkpoint.get('config', {})
        self.hidden_dim = int(config.get('hidden_dim', 768))
        self.num_fusion_layers = int(config.get('num_fusion_layers', 3))
        encoder_checkpoint = config.get('encoder_checkpoint', 'google/vit-base-patch16-224-in21k')

        # Vision encoder (pretrained backbone)
        self.vision_encoder = AutoModel.from_pretrained(encoder_checkpoint)
        encoder_dim = self.vision_encoder.config.hidden_size

        # Projection layers (to phase-1's hidden_dim)
        self.frontal_projection = nn.Linear(encoder_dim, self.hidden_dim)
        self.lateral_projection = nn.Linear(encoder_dim, self.hidden_dim)

        # Fusion layers
        self.fusion_layers = nn.ModuleList([
            BidirectionalCrossAttention(self.hidden_dim) for _ in range(self.num_fusion_layers)
        ])

        # Pooling
        self.pooling = nn.AdaptiveAvgPool1d(1)

        # Load provided weights if present (wrap in try/except to be robust)
        try:
            if 'vision_encoder' in checkpoint:
                self.vision_encoder.load_state_dict(checkpoint['vision_encoder'])
            if 'frontal_projection' in checkpoint:
                self.frontal_projection.load_state_dict(checkpoint['frontal_projection'])
            if 'lateral_projection' in checkpoint:
                self.lateral_projection.load_state_dict(checkpoint['lateral_projection'])
            if 'fusion_layers' in checkpoint:
                # fusion_layers is a ModuleList state_dict: just load
                self.fusion_layers.load_state_dict(checkpoint['fusion_layers'])
        except Exception as e:
            print("Warning: could not fully load some pretrained weights from checkpoint:", e)

        # Freeze by default (you can unfreeze later if desired)
        self.freeze_vision_components()
        self.to(device)

    def freeze_vision_components(self):
        for p in self.vision_encoder.parameters():
            p.requires_grad = False
        for p in self.frontal_projection.parameters():
            p.requires_grad = False
        for p in self.lateral_projection.parameters():
            p.requires_grad = False
        for p in self.fusion_layers.parameters():
            p.requires_grad = False

    def extract_features(self, frontal_images, lateral_images):
        """
        Args:
            frontal_images / lateral_images: tensors [batch, 3, H, W] on same device as this module
        Returns:
            unified_embedding: [batch, hidden_dim*2]
            frontal_patches: [batch, seq_len, hidden_dim]
            lateral_patches: [batch, seq_len, hidden_dim]
        """
        device = next(self.parameters()).device
        frontal_images = frontal_images.to(device)
        lateral_images = lateral_images.to(device)

        with torch.no_grad():
            frontal_outputs = self.vision_encoder(frontal_images, output_hidden_states=False, return_dict=True)
            lateral_outputs = self.vision_encoder(lateral_images, output_hidden_states=False, return_dict=True)

            # last_hidden_state shape: [batch, seq_len, encoder_dim]
            frontal_patches = self.frontal_projection(frontal_outputs.last_hidden_state)
            lateral_patches = self.lateral_projection(lateral_outputs.last_hidden_state)

            # Fusion stack (still in eval/no-grad)
            for fusion_layer in self.fusion_layers:
                frontal_patches, lateral_patches = fusion_layer(frontal_patches, lateral_patches)

            frontal_pooled = self.pooling(frontal_patches.transpose(1, 2)).squeeze(-1)
            lateral_pooled = self.pooling(lateral_patches.transpose(1, 2)).squeeze(-1)
            unified_embedding = torch.cat([frontal_pooled, lateral_pooled], dim=1)

        return unified_embedding, frontal_patches, lateral_patches


# -------------------------
# Visual -> LLM mapper
# -------------------------
class VisualPrefixMapper(nn.Module):
    """
    Map visual embeddings (unified) to a set of prefix token embeddings the LLM can consume.
    """
    def __init__(self, visual_dim, llm_dim, num_prefix_tokens=8):
        super().__init__()
        self.num_prefix_tokens = num_prefix_tokens
        self.mapper = nn.Sequential(
            nn.Linear(visual_dim, llm_dim * 2),
            nn.LayerNorm(llm_dim * 2),
            nn.GELU(),
            nn.Dropout(0.1),
            nn.Linear(llm_dim * 2, llm_dim * num_prefix_tokens),
        )
        self.visual_queries = nn.Parameter(torch.randn(1, num_prefix_tokens, llm_dim))

    def forward(self, visual_embedding):
        # visual_embedding: [batch, visual_dim]
        batch = visual_embedding.shape[0]
        mapped = self.mapper(visual_embedding)  # [batch, llm_dim * num_tokens]
        prefix = mapped.view(batch, self.num_prefix_tokens, -1)  # [batch, num_tokens, llm_dim]
        prefix = prefix + self.visual_queries.expand(batch, -1, -1).to(prefix.device)
        return prefix


# -------------------------
# MultiView CXR -> LLM
# -------------------------
class MultiViewCXRReportGenerator(nn.Module):
    """
    Combines frozen Phase-1 vision backbone + visual prefix mapper + BioGPT (with optional LoRA)
    """
    def __init__(self, vision_checkpoint_path, llm_name="microsoft/biogpt", device='cuda'):
        super().__init__()
        self.device = torch.device(device)

        # Load Phase-1 vision backbone
        self.vision_encoder = PretrainedVisionEncoder(vision_checkpoint_path, device=self.device)
        visual_dim = int(self.vision_encoder.hidden_dim * 2)

        # Tokenizer
        self.tokenizer = AutoTokenizer.from_pretrained(llm_name)
        if self.tokenizer.pad_token is None:
            # ensure a pad token exists
            if self.tokenizer.eos_token is None:
                self.tokenizer.add_special_tokens({"pad_token": "[PAD]"})
            else:
                self.tokenizer.pad_token = self.tokenizer.eos_token

        # Load LLM (BioGPT) - try quantized path if possible
        # Note: BitsAndBytesConfig availability is checked above.
        use_quant = (_HAS_BNB_CONFIG and ("bitsandbytes" in globals() or "bitsandbytes" in str(_HAS_BNB_CONFIG)))
        try:
            if _HAS_BNB_CONFIG:
                # If BitsAndBytesConfig is importable, attempt 4-bit load. This may fail if bitsandbytes is not installed.
                bnb_config = BitsAndBytesConfig(
                    load_in_4bit=True,
                    bnb_4bit_quant_type="nf4",
                    bnb_4bit_compute_dtype=torch.float16,
                    bnb_4bit_use_double_quant=True
                )
                # load (the call below may raise if bitsandbytes not installed)
                self.llm = BioGptForCausalLM.from_pretrained(llm_name, quantization_config=bnb_config, device_map="auto", trust_remote_code=True)
                self._quantized = True
            else:
                raise RuntimeError("BitsAndBytesConfig unavailable")
        except Exception:
            # Fallback: load normally in fp16 if possible (and move to device)
            print("bitsandbytes quantization not used or failed — loading standard fp16 model. "
                  "If you want quantization, install bitsandbytes and compatible transformers.")
            self.llm = BioGptForCausalLM.from_pretrained(llm_name, torch_dtype=torch.float16)
            self.llm.to(self.device)
            self._quantized = False

        # Prepare for k-bit training if available
        if _HAS_PEFT:
            try:
                self.llm = prepare_model_for_kbit_training(self.llm)
            except Exception:
                # prepare_model_for_kbit_training may be a no-op depending on availability
                pass

            # Choose LoRA target modules broadly (these names cover common linear layers across GPT-style models)
            lora_targets = ["q_proj", "k_proj", "v_proj", "o_proj", "c_proj", "out_proj", "fc", "wte", "wo"]

            lora_config = LoraConfig(
                r=16,
                lora_alpha=32,
                target_modules=[t for t in lora_targets],  # some names may not exist; PEFT handles that
                lora_dropout=0.1,
                bias="none",
                task_type=TaskType.CAUSAL_LM
            )

            try:
                self.llm = get_peft_model(self.llm, lora_config)
            except Exception as e:
                print("Warning: LoRA wrapping failed or partially applied:", e)
        else:
            print("Warning: PEFT (LoRA) not available. Continue without LoRA fine-tuning.")

        # Visual mapper & tokens
        llm_dim = getattr(self.llm.config, "hidden_size", 768)
        self.visual_mapper = VisualPrefixMapper(visual_dim, llm_dim, num_prefix_tokens=8).to(self.device)

        self.visual_token = "<VISUAL>"
        self.report_token = "<REPORT>"

        # Move everything to device
        self.to(self.device)

    def prepare_inputs(self, frontal_images, lateral_images, reports=None):
        """
        If reports provided => training mode returns inputs_embeds, attention_mask, labels
        else => inference mode returns visual_prefix embeddings
        """
        batch_size = frontal_images.shape[0]
        device = self.device

        # Extract visual features
        visual_embeddings, _, _ = self.vision_encoder.extract_features(frontal_images, lateral_images)
        # visual_embeddings: [batch, visual_dim]
        visual_prefix = self.visual_mapper(visual_embeddings)  # [batch, num_prefix, llm_dim]
        visual_prefix = visual_prefix.to(device)

        if reports is not None:
            # Training mode
            # Build text sequences: include visual and report tokens so the model can attend to visual prefix
            report_texts = [f"{self.visual_token} {self.report_token} {r}" for r in reports]
            encoded = self.tokenizer(report_texts, padding=True, truncation=True, max_length=512, return_tensors="pt")
            input_ids = encoded["input_ids"].to(device)           # [batch, seq_len]
            attention_mask = encoded["attention_mask"].to(device) # [batch, seq_len]

            # Get token embeddings from the LLM's input embedding module
            input_embeds = self.llm.get_input_embeddings()(input_ids)  # [batch, seq_len, llm_dim]

            # Now build combined embeddings: [visual_prefix tokens] + [text embeddings]
            # Visual prefix length is constant; text seq_len is constant due to tokenizer padding -> can stack
            combined_embeddings = torch.cat([visual_prefix, input_embeds], dim=1)  # [batch, prefix + seq_len, llm_dim]

            # Combined attention mask: prefix tokens attended (1s) + original attention mask
            prefix_attn = torch.ones((batch_size, visual_prefix.shape[1]), dtype=attention_mask.dtype, device=device)
            combined_attention = torch.cat([prefix_attn, attention_mask], dim=1)  # [batch, prefix + seq_len]

            # Prepare labels: shift labels so loss is computed only on the text tokens (not on prefix)
            labels = input_ids.clone()
            # ignore pad tokens for loss
            labels[labels == self.tokenizer.pad_token_id] = -100

            # Build shifted labels of full length (prefix positions -> -100)
            full_seq_len = visual_prefix.shape[1] + input_ids.shape[1]
            shifted_labels = torch.full((batch_size, full_seq_len), -100, dtype=torch.long, device=device)
            shifted_labels[:, visual_prefix.shape[1]:visual_prefix.shape[1] + input_ids.shape[1]] = labels

            return {
                "inputs_embeds": combined_embeddings,
                "attention_mask": combined_attention,
                "labels": shifted_labels
            }
        else:
            # Inference mode: return visual prefix embeddings (batch, prefix_len, llm_dim)
            return visual_prefix

    def generate_report(self, frontal_images, lateral_images, max_length=256, num_beams=4):
        """
        Generate a report given images. This function uses visual prefix + a prompt token,
        and runs the model.generate with inputs_embeds.
        """
        self.eval()
        device = self.device
        with torch.no_grad():
            visual_prefix = self.prepare_inputs(frontal_images, lateral_images, reports=None)  # [batch, prefix, dim]
            # We'll generate for batch-size = 1 in typical usage (but handle batch)
            # Create a prompt that begins generation (report token)
            prompt = f"{self.report_token}"
            prompt_ids = self.tokenizer(prompt, return_tensors="pt", add_special_tokens=False)["input_ids"].to(device)  # [1, L]
            prompt_embeds = self.llm.get_input_embeddings()(prompt_ids)  # [1, L, dim]
            # Repeat prompt for batch
            batch_size = visual_prefix.shape[0]
            prompt_embeds = prompt_embeds.expand(batch_size, -1, -1).to(device)

            # Combined embeddings: visual prefix + prompt
            combined_embeds = torch.cat([visual_prefix, prompt_embeds], dim=1)  # [batch, prefix + prompt_len, dim]

            # Generate (use generate with inputs_embeds)
            outputs = self.llm.generate(
                inputs_embeds=combined_embeds,
                max_length=visual_prefix.shape[1] + prompt_embeds.shape[1] + max_length,
                num_beams=num_beams,
                do_sample=True,
                temperature=0.7,
                top_p=0.9,
                pad_token_id=self.tokenizer.pad_token_id,
                eos_token_id=self.tokenizer.eos_token_id
            )

            # Decode (remove prefix/prompt tokens if present)
            generated_texts = []
            for out in outputs:
                txt = self.tokenizer.decode(out, skip_special_tokens=True)
                # Find report_token and return text after it
                if self.report_token in txt:
                    txt = txt.split(self.report_token)[-1].strip()
                generated_texts.append(txt)
            return generated_texts if len(generated_texts) > 1 else generated_texts[0]


# -------------------------
# Dataset for generation
# -------------------------
class MultiViewReportDataset(Dataset):
    def __init__(self, df, feature_extractor, max_length=512):
        self.df = df.reset_index(drop=True)
        self.feature_extractor = feature_extractor
        self.max_length = max_length

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        frontal_img = Image.open(row["frontal_path"]).convert("RGB")
        lateral_img = Image.open(row["lateral_path"]).convert("RGB")

        frontal_inputs = self.feature_extractor(images=frontal_img, return_tensors="pt")
        lateral_inputs = self.feature_extractor(images=lateral_img, return_tensors="pt")

        return {
            "frontal_pixel_values": frontal_inputs["pixel_values"].squeeze(0),
            "lateral_pixel_values": lateral_inputs["pixel_values"].squeeze(0),
            "report": row["findings"]
        }


# -------------------------
# Evaluator (BLEU / ROUGE / BERTScore)
# -------------------------
class ReportEvaluator:
    def __init__(self, device=None):
        self.bleu_smooth = SmoothingFunction().method4
        # HF evaluate metrics
        self.rouge = evaluate.load("rouge")
        self.bertscore = evaluate.load("bertscore")
        self.device = device if device is not None else ("cuda" if torch.cuda.is_available() else "cpu")

    def evaluate_batch(self, generated_reports, reference_reports):
        # BLEU per-sample and average
        bleu1, bleu2, bleu3, bleu4 = [], [], [], []

        for gen, ref in zip(generated_reports, reference_reports):
            gen_tokens = gen.lower().split()
            ref_tokens = ref.lower().split()
            bleu1.append(sentence_bleu([ref_tokens], gen_tokens, weights=(1, 0, 0, 0), smoothing_function=self.bleu_smooth))
            bleu2.append(sentence_bleu([ref_tokens], gen_tokens, weights=(0.5, 0.5, 0, 0), smoothing_function=self.bleu_smooth))
            bleu3.append(sentence_bleu([ref_tokens], gen_tokens, weights=(0.33, 0.33, 0.34, 0), smoothing_function=self.bleu_smooth))
            bleu4.append(sentence_bleu([ref_tokens], gen_tokens, weights=(0.25, 0.25, 0.25, 0.25), smoothing_function=self.bleu_smooth))

        # ROUGE (batch)
        rouge_results = self.rouge.compute(predictions=generated_reports, references=reference_reports)
        # select rougeL fmeasure if available
        rouge_l = rouge_results.get("rougeL", None)
        if isinstance(rouge_l, dict):
            # old API may return dict with 'fmeasure' or 'f'
            rouge_l_score = rouge_l.get("fmeasure", rouge_l.get("f", 0.0))
        else:
            rouge_l_score = rouge_l

        # BERTScore (batch)
        bert = self.bertscore.compute(predictions=generated_reports, references=reference_reports, lang="en", device=self.device)
        bert_f1_mean = float(np.mean(bert["f1"])) if "f1" in bert else 0.0

        # Averages
        return {
            "bleu-1": float(np.mean(bleu1)) if bleu1 else 0.0,
            "bleu-2": float(np.mean(bleu2)) if bleu2 else 0.0,
            "bleu-3": float(np.mean(bleu3)) if bleu3 else 0.0,
            "bleu-4": float(np.mean(bleu4)) if bleu4 else 0.0,
            "rouge-L": float(rouge_l_score) if rouge_l_score is not None else 0.0,
            "bertscore-f1": bert_f1_mean,
        }


# -------------------------
# Data preparation helpers
# -------------------------
def prepare_data_for_generation(df1, df2, image_dir):
    generation_data = []
    patient_groups = df2.groupby("uid")
    for uid, group in patient_groups:
        report_row = df1[df1["uid"] == uid]
        if report_row.empty:
            continue
        findings = report_row["findings"].values[0]
        impression = report_row["impression"].values[0] if "impression" in report_row.columns else ""
        if pd.isna(findings):
            continue
        full_report = f"FINDINGS: {findings}"
        if pd.notna(impression) and impression.strip():
            full_report += f"\nIMPRESSION: {impression}"

        frontal_imgs = group[group["projection"].str.contains("Frontal", case=False, na=False)]
        lateral_imgs = group[group["projection"].str.contains("Lateral", case=False, na=False)]
        if frontal_imgs.empty or lateral_imgs.empty:
            continue
        frontal_path = os.path.join(image_dir, frontal_imgs.iloc[0]["filename"])
        lateral_path = os.path.join(image_dir, lateral_imgs.iloc[0]["filename"])
        generation_data.append({
            "uid": uid,
            "frontal_path": frontal_path,
            "lateral_path": lateral_path,
            "findings": full_report
        })
    return pd.DataFrame(generation_data)


# -------------------------
# Training loop for report generator
# -------------------------
def train_report_generator(model, train_loader, val_loader, num_epochs=3, learning_rate=2e-4):
    trainable_params = [p for p in model.parameters() if p.requires_grad]
    optimizer = torch.optim.AdamW(trainable_params, lr=learning_rate)
    model.train()
    train_losses, val_losses = [], []

    for epoch in range(num_epochs):
        print(f"\nEpoch {epoch+1}/{num_epochs}")
        epoch_loss = 0.0
        pbar = tqdm(train_loader, desc="Training")
        for batch in pbar:
            frontal = batch["frontal_pixel_values"].to(model.device)
            lateral = batch["lateral_pixel_values"].to(model.device)
            reports = list(batch["report"])

            inputs = model.prepare_inputs(frontal, lateral, reports)
            outputs = model.llm(
                inputs_embeds=inputs["inputs_embeds"].to(model.device),
                attention_mask=inputs["attention_mask"].to(model.device),
                labels=inputs["labels"].to(model.device),
            )
            loss = outputs.loss
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item()
            pbar.set_postfix({"loss": float(loss.item())})

        avg_train_loss = epoch_loss / max(1, len(train_loader))
        train_losses.append(avg_train_loss)

        # Validation
        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for batch in tqdm(val_loader, desc="Validation"):
                frontal = batch["frontal_pixel_values"].to(model.device)
                lateral = batch["lateral_pixel_values"].to(model.device)
                reports = list(batch["report"])
                inputs = model.prepare_inputs(frontal, lateral, reports)
                outputs = model.llm(
                    inputs_embeds=inputs["inputs_embeds"].to(model.device),
                    attention_mask=inputs["attention_mask"].to(model.device),
                    labels=inputs["labels"].to(model.device),
                )
                val_loss += float(outputs.loss.item())

        avg_val_loss = val_loss / max(1, len(val_loader))
        val_losses.append(avg_val_loss)
        print(f"Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f}")
        model.train()

    return train_losses, val_losses


# -------------------------
# Main script
# -------------------------
def main():
    # Configuration
    config = {
        "vision_checkpoint": "/kaggle/working/multiview_vision_backbone.pth",
        "llm_name": "microsoft/biogpt",
        "batch_size": 2,
        "learning_rate": 2e-4,
        "num_epochs": 2,
        "device": torch.device("cuda" if torch.cuda.is_available() else "cpu"),
    }

    print("Phase 2: Report Generation")
    print("Device:", config["device"])

    # Load IU dataset
    df1 = pd.read_csv("/kaggle/input/chest-xrays-indiana-university/indiana_reports.csv")
    df2 = pd.read_csv("/kaggle/input/chest-xrays-indiana-university/indiana_projections.csv")
    image_dir = "/kaggle/input/chest-xrays-indiana-university/images/images_normalized/"

    generation_df = prepare_data_for_generation(df1, df2, image_dir)
    print(f"Found {len(generation_df)} pairs with reports")

    train_df, val_df = train_test_split(generation_df, test_size=0.15, random_state=42)
    print("Train / Val:", len(train_df), len(val_df))

    # Feature extractor (same as Phase 1 encoder)
    feature_extractor = AutoFeatureExtractor.from_pretrained("google/vit-base-patch16-224-in21k")

    train_dataset = MultiViewReportDataset(train_df, feature_extractor)
    val_dataset = MultiViewReportDataset(val_df, feature_extractor)

    train_loader = DataLoader(train_dataset, batch_size=config["batch_size"], shuffle=True, num_workers=2, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=config["batch_size"], shuffle=False, num_workers=2, pin_memory=True)

    # Initialize model
    print("Initializing model (this may take time)...")
    model = MultiViewCXRReportGenerator(
        vision_checkpoint_path=config["vision_checkpoint"],
        llm_name=config["llm_name"],
        device=config["device"]
    )

    # Parameter counts
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Total params: {total_params:,} | Trainable: {trainable_params:,}")

    # Train
    train_losses, val_losses = train_report_generator(model, train_loader, val_loader, num_epochs=config["num_epochs"], learning_rate=config["learning_rate"])

    # Save model artifacts
    save_path = "/kaggle/working/multiview_report_generator.pth"
    print("Saving artifacts to", save_path)
    try:
        torch.save({
            "visual_mapper": model.visual_mapper.state_dict(),
            "llm_state": model.llm.state_dict(),
            "config": config
        }, save_path)
    except Exception as e:
        print("Warning: saving failed with exception:", e)
        # as fallback try saving only peft weights if available
        try:
            model.llm.save_pretrained("/kaggle/working/multiview_report_generator_peft")
            print("Saved peft model to /kaggle/working/multiview_report_generator_peft")
        except Exception as e2:
            print("Fallback save also failed:", e2)

    # Generate a few sample reports from validation set
    model.eval()
    evaluator = ReportEvaluator(device=str(config["device"]))
    sample_count = min(5, len(val_dataset))
    if sample_count == 0:
        print("No validation samples to generate.")
        return

    sample_indices = np.random.choice(len(val_dataset), sample_count, replace=False)
    generated_reports, reference_reports = [], []
    print("\nGenerating sample reports:")
    for idx in sample_indices:
        sample = val_dataset[idx]
        frontal = sample["frontal_pixel_values"].unsqueeze(0).to(config["device"])
        lateral = sample["lateral_pixel_values"].unsqueeze(0).to(config["device"])
        reference = sample["report"]
        generated = model.generate_report(frontal, lateral, max_length=128)
        print("-" * 60)
        print("Reference (truncated):", reference[:240].replace("\n", " ") + "...")
        print("Generated (truncated):", (generated[:240] if isinstance(generated, str) else str(generated))[:240] + "...")
        generated_reports.append(generated if isinstance(generated, str) else generated)
        reference_reports.append(reference)

    metrics = evaluator.evaluate_batch(generated_reports, reference_reports)
    print("\nEvaluation metrics:")
    for k, v in metrics.items():
        print(f"{k}: {v:.4f}")

    # Plot training curves & architecture summary
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.plot(train_losses, label="Train Loss")
    plt.plot(val_losses, label="Val Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()
    plt.title("Training Curves")

    plt.subplot(1, 2, 2)
    plt.axis("off")
    plt.text(0.5, 0.5, "Vision Encoder (Frozen)\n↓\nCross-Attention Fusion\n↓\nVisual Prefix Mapper\n↓\nBioGPT (+LoRA if available)\n↓\nGenerated Report", ha="center", va="center")

    plt.tight_layout()
    plt.savefig("/kaggle/working/phase2_training.png")
    plt.show()

    print("\nPhase 2 completed. Model artifacts saved where possible.")


if __name__ == "__main__":
    main()
```

```python
import torch
import pandas as pd
from PIL import Image
import os
import warnings
from tqdm import tqdm
import numpy as np
from torch.utils.data import DataLoader

# --- Reuse necessary classes and functions from your project ---
# Make sure the following are defined in your environment:
# ReportEvaluator, MultiViewCXRReportGenerator, MultiViewReportDataset,
# prepare_data_for_generation, AutoFeatureExtractor.
# For simplicity, the evaluator class is included below.

# --- Suppress warnings for cleaner output ---
warnings.filterwarnings('ignore')

# ================== EVALUATOR CLASS ==================

class ReportEvaluator:
    """Calculates BLEU, ROUGE, and BERTScore for generated reports."""
    def __init__(self, device=None):
        self.bleu_smooth = SmoothingFunction().method4
        try:
            import evaluate
            self.rouge = evaluate.load("rouge")
            self.bertscore = evaluate.load("bertscore")
        except ImportError:
            raise ImportError("Please install the 'evaluate' and 'bert_score' libraries from Hugging Face: pip install evaluate bert_score")
            
        self.device = device if device is not None else ("cuda" if torch.cuda.is_available() else "cpu")
        print(f"Evaluator using device: {self.device}")

    def evaluate_batch(self, generated_reports, reference_reports):
        # Ensure inputs are lists of strings
        if not all(isinstance(r, str) for r in generated_reports):
            print("Warning: Some generated items are not strings. Converting them.")
            generated_reports = [str(r) for r in generated_reports]
        if not all(isinstance(r, str) for r in reference_reports):
            print("Warning: Some reference items are not strings. Converting them.")
            reference_reports = [str(r) for r in reference_reports]

        # --- BLEU Scores ---
        bleu_scores = {'bleu-1': [], 'bleu-2': [], 'bleu-3': [], 'bleu-4': []}
        for gen, ref in zip(generated_reports, reference_reports):
            gen_tokens = gen.lower().split()
            ref_tokens = [ref.lower().split()]
            for i in range(1, 5):
                weights = tuple(1/i for _ in range(i)) + tuple(0 for _ in range(4-i))
                bleu_scores[f'bleu-{i}'].append(sentence_bleu(ref_tokens, gen_tokens, weights=weights, smoothing_function=self.bleu_smooth))

        # --- ROUGE Score ---
        rouge_results = self.rouge.compute(predictions=generated_reports, references=reference_reports)
        
        # --- BERTScore ---
        bert_results = self.bertscore.compute(predictions=generated_reports, references=reference_reports, lang="en", device=self.device)
        
        # --- Compile and return average scores ---
        final_metrics = {k: np.mean(v) for k, v in bleu_scores.items()}
        final_metrics['rouge-L'] = rouge_results.get('rougeL', 0.0)
        final_metrics['bertscore-f1'] = np.mean(bert_results['f1'])
        
        return final_metrics

# ================== MAIN EVALUATION FUNCTION ==================

def evaluate_overall_performance(model, dataloader, evaluator):
    """
    Generates reports for an entire dataset and computes overall scores.
    """
    model.eval()
    all_generated_reports = []
    all_reference_reports = []
    
    with torch.no_grad():
        # Use tqdm for a progress bar
        for batch in tqdm(dataloader, desc="Generating reports for validation set"):
            frontal_images = batch['frontal_pixel_values'].to(model.device)
            lateral_images = batch['lateral_pixel_values'].to(model.device)
            reference_reports = batch['report'] # This is a tuple of strings
            
            # Generate reports for the batch
            generated_reports = model.generate_report(
                frontal_images=frontal_images,
                lateral_images=lateral_images,
                max_length=256,
                num_beams=4
            )
            
            # The generate function might return a single string for a batch size of 1
            if isinstance(generated_reports, str):
                generated_reports = [generated_reports]

            # Clean the generated reports (remove repetitive tails)
            cleaned_reports = []
            for report in generated_reports:
                # Find the first sign of repetition and cut it off
                if "..." in report:
                    report = report.split("...")[0]
                cleaned_reports.append(report)

            all_generated_reports.extend(cleaned_reports)
            all_reference_reports.extend(reference_reports)
            
    print("\nGeneration complete. Calculating overall scores...")
    
    # Calculate scores on all collected reports
    overall_scores = evaluator.evaluate_batch(all_generated_reports, all_reference_reports)
    return overall_scores


# ================== SCRIPT EXECUTION ==================

# --- Configuration ---
PHASE_1_CHECKPOINT = '/kaggle/working/multiview_vision_backbone.pth'
PHASE_2_CHECKPOINT = '/kaggle/working/multiview_report_generator.pth'
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# --- Load Model (ensure MultiViewCXRReportGenerator is defined) ---
print("Loading the trained report generation model...")
model = MultiViewCXRReportGenerator(
    vision_checkpoint_path=PHASE_1_CHECKPOINT,
    llm_name="microsoft/biogpt",
    device=DEVICE
)
state_dict = torch.load(PHASE_2_CHECKPOINT, map_location=DEVICE)
model.visual_mapper.load_state_dict(state_dict['visual_mapper'])
model.llm.load_state_dict(state_dict['llm_state'], strict=False)
print("Model loaded successfully.")

# --- Load Data (ensure data prep functions and classes are defined) ---
print("Loading and preparing validation data...")
df1 = pd.read_csv('/kaggle/input/chest-xrays-indiana-university/indiana_reports.csv')
df2 = pd.read_csv('/kaggle/input/chest-xrays-indiana-university/indiana_projections.csv')
image_dir = '/kaggle/input/chest-xrays-indiana-university/images/images_normalized/'

generation_df = prepare_data_for_generation(df1, df2, image_dir)
_, val_df = train_test_split(generation_df, test_size=0.15, random_state=42)

feature_extractor = AutoFeatureExtractor.from_pretrained('google/vit-base-patch16-224-in21k')
val_dataset = MultiViewReportDataset(val_df, feature_extractor)
val_loader = DataLoader(val_dataset, batch_size=4, shuffle=False) # Use a reasonable batch size for inference

# --- Run Evaluation ---
evaluator = ReportEvaluator(device=DEVICE)
final_scores = evaluate_overall_performance(model, val_loader, evaluator)

# --- Display Final Scores ---
print("\n" + "="*50)
print("      OVERALL MODEL PERFORMANCE (VALIDATION SET)")
print("="*50)
for metric, score in final_scores.items():
    print(f"{metric:<15}: {score:.4f}")
print("="*50)
```

```python
import torch
import pandas as pd
from PIL import Image
import os
import warnings
import random
from transformers import AutoFeatureExtractor

# --- Suppress warnings for cleaner output ---
warnings.filterwarnings('ignore')


def test_random_uids(model, reports_df, projections_df, image_dir, num_samples=3):
    """
    Selects random UIDs, generates reports, and compares them to the ground truth.

    Args:
        model: The trained MultiViewCXRReportGenerator model.
        reports_df (pd.DataFrame): DataFrame with report findings (indiana_reports.csv).
        projections_df (pd.DataFrame): DataFrame with image paths (indiana_projections.csv).
        image_dir (str): The directory where images are stored.
        num_samples (int): The number of random samples to test.
    """
    model.eval()
    
    # Get a list of unique UIDs that have associated images
    available_uids = projections_df['uid'].unique().tolist()
    
    # Select random UIDs from the list
    if len(available_uids) < num_samples:
        print(f"Warning: Requested {num_samples} samples, but only {len(available_uids)} UIDs are available.")
        num_samples = len(available_uids)
        
    random_uids_to_test = random.sample(available_uids, num_samples)
    print(f"Selected random UIDs for testing: {random_uids_to_test}\n")

    # Initialize the feature extractor
    feature_extractor = AutoFeatureExtractor.from_pretrained('google/vit-base-patch16-224-in21k')
    
    for uid in random_uids_to_test:
        print(f"\n{'='*80}")
        print(f"Testing UID: {uid}")
        print(f"{'-'*80}")

        # --- 1. Find the Ground Truth Report ---
        report_info = reports_df[reports_df['uid'] == uid]
        if report_info.empty:
            print(f"--> No report found for UID {uid}. Skipping.")
            continue
        
        findings = report_info.iloc[0]['findings']
        impression = report_info.iloc[0]['impression']
        ground_truth_report = f"FINDINGS: {findings}\nIMPRESSION: {impression}"

        # --- 2. Find the Frontal and Lateral Images ---
        image_info = projections_df[projections_df['uid'] == uid]
        frontal_path_info = image_info[image_info['projection'].str.contains("Frontal", case=False, na=False)]
        lateral_path_info = image_info[image_info['projection'].str.contains("Lateral", case=False, na=False)]

        if frontal_path_info.empty or lateral_path_info.empty:
            print(f"--> Could not find a frontal/lateral pair for UID {uid}. Skipping.")
            continue
            
        frontal_path = os.path.join(image_dir, frontal_path_info.iloc[0]['filename'])
        lateral_path = os.path.join(image_dir, lateral_path_info.iloc[0]['filename'])

        # --- 3. Load, Process, and Generate ---
        try:
            frontal_img = Image.open(frontal_path).convert("RGB")
            lateral_img = Image.open(lateral_path).convert("RGB")
        except FileNotFoundError:
            print(f"--> Image file not found for UID {uid}. Skipping.")
            continue

        frontal_pixel_values = feature_extractor(images=frontal_img, return_tensors="pt")['pixel_values'].to(model.device)
        lateral_pixel_values = feature_extractor(images=lateral_img, return_tensors="pt")['pixel_values'].to(model.device)

        print("Generating report...")
        with torch.no_grad():
            generated_report = model.generate_report(
                frontal_images=frontal_pixel_values,
                lateral_images=lateral_pixel_values,
                max_length=256,
                num_beams=5
            )

        # --- 4. Display Comparison ---
        print("\n--- GROUND TRUTH REPORT ---\n")
        print(ground_truth_report)
        print("\n\n--- MODEL GENERATED REPORT ---\n")
        print(generated_report)

    print(f"\n{'='*80}")
    print("Testing complete.")


# ================== SCRIPT EXECUTION ==================

# --- Configuration ---
PHASE_1_CHECKPOINT = '/kaggle/working/multiview_vision_backbone.pth'
PHASE_2_CHECKPOINT = '/kaggle/working/multiview_report_generator.pth'
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
REPORTS_CSV_PATH = '/kaggle/input/chest-xrays-indiana-university/indiana_reports.csv'
PROJECTIONS_CSV_PATH = '/kaggle/input/chest-xrays-indiana-university/indiana_projections.csv'
IMAGE_DIR_PATH = '/kaggle/input/chest-xrays-indiana-university/images/images_normalized/'

# --- Load Model (ensure MultiViewCXRReportGenerator is defined in your environment) ---
print("Loading the trained report generation model...")
model = MultiViewCXRReportGenerator(
    vision_checkpoint_path=PHASE_1_CHECKPOINT,
    llm_name="microsoft/biogpt",
    device=DEVICE
)
# Load the fine-tuned weights
state_dict = torch.load(PHASE_2_CHECKPOINT, map_location=DEVICE)
model.visual_mapper.load_state_dict(state_dict['visual_mapper'])
model.llm.load_state_dict(state_dict['llm_state'], strict=False)
print("Model loaded successfully.")

# --- Load DataFrames ---
reports_df = pd.read_csv(REPORTS_CSV_PATH)
projections_df = pd.read_csv(PROJECTIONS_CSV_PATH)

# --- Run the Test ---
# You can change the number of samples to test here
test_random_uids(model, reports_df, projections_df, IMAGE_DIR_PATH, num_samples=3)
```
